```python
from pathlib import Path

file = Path(r"C:\Users\maazs\OneDrive\Desktop\PricesDesk\teaching_prices_clean.csv")

print("CSV found:", file.exists())
```

```python
import csv

with open(file, "r", encoding="utf-8-sig", newline="") as f:
    reader = csv.DictReader(f)
    data = list(reader)
    columns = reader.fieldnames

print("Total rows:", len(data))
print("Total columns:", len(columns))
print("Column names:", columns)

print("\nFirst 3 records:")
for row in data[:3]:
    print(row)
```

```python
unique_items = set(row["item"] for row in data)
unique_dates = set(row["date"] for row in data)

print("Number of distinct items:", len(unique_items))
print("Number of distinct dates:", len(unique_dates))

print("\nItems in dataset:")
print(sorted(unique_items))

print("\nDates in dataset:")
print(sorted(unique_dates))
```

```python

print("PREDICTION VS ACTUAL DATA")
print("-------------------------")

print("Rows: Predicted 250, Actual", len(data))
print("Columns: Predicted 4, Actual", len(columns))
print("Items: Predicted 10, Actual", len(unique_items))
print("Dates: Predicted 25, Actual", len(unique_dates))

print("\nCOMPARISON:")
print("The dataset has fewer rows than predicted.")
print("The dataset has more columns than predicted.")
print("There are fewer distinct items and dates than predicted.")
```

```python

print("DATASET DETAILS")
print("----------------")

for column in ["item", "city", "unit", "data_label"]:
    values = sorted(set(row[column] for row in data))
    print(column, ":", values)

print("\nMISSING VALUES")
for column in columns:
    missing = sum(
        1 for row in data
        if not row[column].strip()
    )
    print(column, ":", missing)

print("\nDATE RANGE")
print("First date:", min(unique_dates))
print("Last date:", max(unique_dates))
```
